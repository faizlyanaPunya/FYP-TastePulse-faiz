```python
import pandas as pd
import numpy as np

# Load your dataset (e.g., CSV file)
df = pd.read_csv('TPP_Place.csv')

# Identify missing values 
print("Missing data count:\n", df.isnull().sum()) 
```

```python
import malaya
print(malaya.__version__)
```

```python
model = malaya.sentiment.huggingface()
```

```python
malaya.sentiment.label
```

```python
df["text_clean"].apply(type).value_counts()
```

```python
df["text_clean"] = (
    df["text_clean"]
    .astype(str)          # convert everything to string
    .str.strip()          # remove empty spaces
)

# remove empty / invalid texts
df = df[df["text_clean"].str.len() > 0]
```

```python
df["text_clean"].apply(type).value_counts()
```

```python
batch_size = 16  # reduce to 8 if RAM is small

sentiments = []

for i in range(0, len(df), batch_size):
    batch = df["text_clean"].iloc[i:i+batch_size].astype(str).tolist()
    preds = model.predict(batch)
    sentiments.extend(preds)

df["sentiment"] = sentiments
```

```python
df[["text_clean", "sentiment"]].head()
```

```python
#save to csv
df.to_csv("sentiment_results.csv", index=False, encoding="utf-8-sig")
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['label'] = le.fit_transform(df['sentiment'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text_clean'], df['label'], test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Convert text to TF-IDF features
tfidf = TfidfVectorizer(max_features=5000)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# Train ML model
ml_model = MultinomialNB()
ml_model.fit(X_train_tfidf, y_train)

# Predict and evaluate
y_pred_ml = ml_model.predict(X_test_tfidf)
print("ML Model Accuracy:", accuracy_score(y_test, y_pred_ml))
print(classification_report(y_test, y_pred_ml))
```
