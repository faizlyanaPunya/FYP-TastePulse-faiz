```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
```

```python
df = pd.read_csv('Modelvs.csv')  # assuming your CSV file
```

```python
df["text_clean"].apply(type).value_counts()
```

```python
df["text_clean"] = (
    df["text_clean"]
    .astype(str)          # convert everything to string
    .str.strip()          # remove empty spaces
)

# remove empty / invalid texts
df = df[df["text_clean"].str.len() > 0]
```

```python
df["text_clean"].apply(type).value_counts()
```

```python
texts = df['text_clean']
labels = df['label']

# Encode labels (positive=2, neutral=1, negative=0, etc.)
target = LabelEncoder().fit_transform(labels)
```

```python
df["text_clean"].apply(type).value_counts()
```

```python
tfidf = TfidfVectorizer(ngram_range=(1,3), min_df=2).fit(texts)
vectors = tfidf.transform(texts)
print(vectors.shape)
```

```python
train_X, test_X, train_Y, test_Y = train_test_split(vectors, target, test_size=0.2, random_state=42)
```

```python
multinomial = MultinomialNB().fit(train_X, train_Y)
```

```python
from sklearn import metrics

# Define string labels matching your numeric encoding
label_names = ['negative', 'neutral', 'positive']
```

```python
# Training performance
print(metrics.classification_report(train_Y, multinomial.predict(train_X), target_names=label_names))
```

```python
# Test performance
print(metrics.classification_report(test_Y, multinomial.predict(test_X), target_names=label_names))
```
